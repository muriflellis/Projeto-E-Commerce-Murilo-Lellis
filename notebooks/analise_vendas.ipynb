```python
import pandas as pd
```

```python
df_vendas = pd.read_excel('../data/dados_ecommerce.xlsx', sheet_name='vendas_olist')
```

```python
df_vendas['produto_categoria'] = df_vendas['produto_categoria'].fillna('Não Informado')
```

```python
df_vendas['cliente_cidade'] = df_vendas['cliente_cidade'].fillna('Desconhecido')
```

```python
df_vendas = df_vendas.dropna(subset=['preco'])
```

```python
df_vendas.isna().sum()
```

```python
df_vendas.groupby('produto_categoria')["preco"].sum()
```

```python
faturamente_categoria=df_vendas.groupby('produto_categoria')["preco"].sum().sort_values(ascending=False)
```

```python
faturamento_caterogia = faturamente_categoria
faturamento_caterogia
```

```python
df_vendas.groupby('produto_categoria')['preco'].agg(['sum','mean','count']).sort_values(by='sum',ascending=False)
```

```python
analise_vendas = df_vendas.groupby('produto_categoria')['preco'].agg(['sum','mean','count']).sort_values(by='sum',ascending=False)
```

```python
analise_cidade = df_vendas.groupby('cliente_cidade')['preco'].agg(['sum','count']).sort_values(by='sum', ascending=False)
analise_cidade
```
