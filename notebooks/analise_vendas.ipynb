Análise

```python
df_tratado['preco'].dropna().sum()
```

```python
df_tratado['preco'].mean()
```

```python
df_tratado.groupby(['produto_categoria'])['preco'].sum().sort_values(ascending=False)
```

```python
df_tratado['mes'] = df_tratado['data_compra'].dt.month
df_tratado['ano'] = df_tratado['data_compra'].dt.year
```

```python
df_tratado.head()
```

```python
df_tratado.groupby(['ano', 'mes'])['preco'].sum().sort_values(ascending=False)
```

```python
df_tratado['data_compra'].min(), df_tratado['data_compra'].max()
```

```python
vendas_mensais = df_tratado.groupby(['ano','mes'])['preco'].sum().reset_index()
vendas_mensais
```

```python
vendas_mensais['variacao_percentual'] = vendas_mensais['preco'].pct_change()
vendas_mensais
```

```python
plt.figure(figsize=(12, 5))
plt.plot(vendas_mensais['mes'], vendas_mensais['preco'])
plt.title('Faturamento mensal em 2025')
plt.xlabel('Mês')
plt.ylabel('Faturamento (R$)')
plt.xticks(range(1,13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
vendas_categoria = (df_tratado.groupby('produto_categoria')['preco'].sum().reset_index().sort_values(by='preco', ascending=False) .reset_index(drop=True))
vendas_categoria
```

```python
plt.figure(figsize=(12,5))
plt.bar(vendas_categoria['produto_categoria'],vendas_categoria['preco'])
plt.title('Faturamento por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Faturamento (R$)')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
vendas_cidade = (df_tratado.groupby('cliente_cidade')['preco'].sum().reset_index().sort_values(by='preco', ascending=False).reset_index(drop=True))
vendas_cidade
```

```python
plt.Figure(figsize=(15,7))
plt.bar(vendas_cidade['cliente_cidade'], vendas_cidade['preco'])
plt.title('Faturamento por Cidade')
plt.xlabel('Cidades')
plt.ylabel('Faturamento (R$)')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
distribuicao_avaliacao = (df_tratado.groupby('avaliacao_score')['id_pedido'].count().reset_index().sort_values(by='avaliacao_score', ascending=False).reset_index(drop=True))
distribuicao_avaliacao
```

```python
df_tratado['percentual_frete'] = (df_tratado['frete']/df_tratado['preco'])*100
df_tratado[['preco','frete','percentual_frete']].head(10)
```

```python
df_tratado['percentual_frete'].describe()
```

```python
plt.Figure(figsize=(15,5))
plt.hist(df_tratado['percentual_frete'], bins=20)
plt.title('Distribuição do Percentual de Frete')
plt.xlabel('Percentual de Frete %')
plt.ylabel('Quantidade de Pedidos')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
