Tratamento dos Dados

```python
df_tratado = df_vendas.copy()
```

```python
df_tratado.info()
```

```python
df_tratado['data_compra'] = pd.to_datetime(df_tratado['data_compra'], errors="coerce")
```

```python
df_tratado['data_compra'].isna().sum()
```

```python
df_tratado['data_compra'].dtype
```

```python
df_tratado.isna().sum()
```

```python
df_tratado['produto_categoria'] = df_tratado['produto_categoria'].fillna('Não Informado')
df_tratado['cliente_cidade'] = df_tratado['cliente_cidade'].fillna('Não Informado')
```

```python
df_tratado[['produto_categoria','cliente_cidade']].isna().sum()
```

```python
df_tratado.loc[df_tratado['preco'].isna(),['id_pedido', 'produto_categoria','preco','frete']]
```

Análise

```python
df_tratado['preco'].dropna().sum()
```

```python
df_tratado['preco'].mean()
```

```python
df_tratado.groupby(['produto_categoria'])['preco'].sum().sort_values(ascending=False)
```

```python
df_tratado['mes'] = df_tratado['data_compra'].dt.month
df_tratado['ano'] = df_tratado['data_compra'].dt.year
```

```python
df_tratado.head()
```

```python
df_tratado.groupby(['ano', 'mes'])['preco'].sum().sort_values(ascending=False)
```

```python
df_tratado['data_compra'].min(), df_tratado['data_compra'].max()
```

```python
vendas_mensais = df_tratado.groupby(['ano','mes'])['preco'].sum().reset_index()
vendas_mensais
```

```python
vendas_mensais['variacao_percentual'] = vendas_mensais['preco'].pct_change()
vendas_mensais
```
